```python
variables = [
    "Administrative",
    "ProductRelated",
    "ProductRelated_Duration"
]

datos[variables].agg(["mean", "median", "min", "max", "std"]).round(2)
```

### Interpretación del análisis estadístico

Al analizar las tres variables, se encontraron los siguientes resultados:

- **Administrative:** el promedio es de 2.32 páginas y la mediana es de 1. Esto indica que la cantidad de páginas administrativas visitadas generalmente es baja.
- **ProductRelated:** el promedio es de 31.73 páginas y la mediana es de 18. Esto muestra que algunos usuarios visitaron una cantidad mayor de páginas de productos.
- **ProductRelated_Duration:** el promedio es de 1194.75 segundos y la mediana es de 598.94 segundos. Existe una diferencia considerable entre ambos valores, lo que puede indicar que algunas sesiones tuvieron tiempos de navegación mucho más largos.

En general, los resultados muestran diferencias entre los usuarios en cuanto a la cantidad de páginas visitadas y el tiempo que permanecieron navegando.

```python
import matplotlib.pyplot as plt
import seaborn as sns

variables = [
    "Administrative",
    "ProductRelated",
    "ProductRelated_Duration"
]

for variable in variables:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=datos[variable])
    plt.title("Valores atípicos de " + variable)
    plt.xlabel(variable)
    plt.show()
```

```python
variables = [
    "Administrative",
    "ProductRelated",
    "ProductRelated_Duration"
]

for variable in variables:
    Q1 = datos[variable].quantile(0.25)
    Q3 = datos[variable].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.15 * IQR
    limite_superior = Q3 + 1.15 * IQR

    apticos = datos[
       (datos[variable] < limite_inferior) |
       (datos[variable] > limite_superior)
    ]
    print("variable: ", variable)
    print("limite_inferior: ", limite_inferior)
    print("limite_superior: ", limite_superior)
    print("Cantidad de valores apticos: ", len(apticos))
```

### Interpretación de los valores atípicos

Mediante el método del rango intercuartílico (IQR), se identificaron valores atípicos en las tres variables analizadas.

La variable `Administrative` presentó 782 valores atípicos, `ProductRelated` presentó 1,240 y `ProductRelated_Duration` presentó 1,203.

Estos resultados indican que existen sesiones donde los usuarios visitaron una cantidad de páginas o permanecieron navegando durante más tiempo que la mayoría.

Estos valores no necesariamente son errores, ya que pueden representar comportamientos reales de los compradores. Por esta razón, no se eliminarán durante esta actividad. En una etapa posterior se podrá evaluar si es necesario aplicar algún tratamiento.
