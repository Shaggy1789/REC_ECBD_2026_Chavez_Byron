# Exploración y diagnóstico del conjunto de datos

## 1. Introducción

En esta actividad se realizará una exploración del conjunto de datos Online Shoppers Purchasing Intention Dataset, el cual contiene información sobre las visitas de usuarios a un sitio de comercio electrónico.

El propósito es conocer cómo están organizados los datos, identificar sus variables y revisar si existen problemas como valores faltantes, registros duplicados o datos fuera de lo normal.

También se realizarán algunos análisis estadísticos y gráficas para comprender mejor la información antes de comenzar con la limpieza y transformación de los datos.
